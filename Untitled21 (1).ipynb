```python
import random, time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, confusion_matrix, ConfusionMatrixDisplay)

def set_seed(seed=42):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)

SEED = 42
set_seed(SEED)
```

```python
N_SAMPLES, NOISE = 1000, 0.20

X, y = make_moons(n_samples=N_SAMPLES, noise=NOISE, random_state=SEED)

# 1) hold out test set
X_trfull, X_test, y_trfull, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=SEED)
# 2) validation set carved from the training portion
X_train, X_val, y_train, y_val = train_test_split(
    X_trfull, y_trfull, test_size=0.20, stratify=y_trfull, random_state=SEED)

# 3) scaler fitted on TRAINING data only
scaler = StandardScaler().fit(X_train)
Xs_train, Xs_val, Xs_test = (scaler.transform(a).astype(np.float32) for a in (X_train, X_val, X_test))

print("train/val/test:", len(X_train), len(X_val), len(X_test))
plt.figure(figsize=(5, 4))
plt.scatter(X[y == 0, 0], X[y == 0, 1], s=12, label="Class 0")
plt.scatter(X[y == 1, 0], X[y == 1, 1], s=12, label="Class 1")
plt.title("make_moons"); plt.xlabel("x1"); plt.ylabel("x2"); plt.legend(); plt.show()
```

```python
class ANN(nn.Module):
    def __init__(self, hidden_layers=(16, 16), input_dim=2):
        super().__init__()
        if not hidden_layers or any((not isinstance(h, int)) or h < 2 for h in hidden_layers):
            raise ValueError("hidden_layers must be a non-empty list of ints >= 2")
        layers, prev = [], input_dim
        for h in hidden_layers:
            layers += [nn.Linear(prev, h), nn.ReLU()]
            prev = h
        layers.append(nn.Linear(prev, 1))          # 1 logit; sigmoid applied only for prediction
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x).squeeze(-1)

@torch.no_grad()
def predict_proba(model, X_scaled):
    model.eval()
    return torch.sigmoid(model(torch.as_tensor(X_scaled, dtype=torch.float32))).numpy()

print(ANN([16, 16]))
```

```python
def train_model(hidden_layers=(16, 16), lr=0.01, epochs=200, batch_size=32, seed=SEED, verbose=True):
    set_seed(seed)
    model = ANN(list(hidden_layers))
    criterion = nn.BCEWithLogitsLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    Xtr = torch.tensor(Xs_train); ytr = torch.tensor(y_train, dtype=torch.float32)
    Xva = torch.tensor(Xs_val);   yva = torch.tensor(y_val, dtype=torch.float32)
    loader = DataLoader(TensorDataset(Xtr, ytr), batch_size=batch_size, shuffle=True,
                        generator=torch.Generator().manual_seed(seed))

    train_hist, val_hist = [], []
    t0 = time.perf_counter()
    for epoch in range(1, epochs + 1):
        model.train()
        total = 0.0
        for xb, yb in loader:
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)   # forward + loss
            loss.backward()                   # backpropagation
            optimizer.step()                  # Adam update
            total += loss.item() * len(xb)
        train_hist.append(total / len(Xtr))

        model.eval()
        with torch.no_grad():                 # validation: no gradient updates
            val_hist.append(criterion(model(Xva), yva).item())

        if verbose and (epoch % 20 == 0 or epoch == 1):
            print(f"epoch {epoch:4d} | train {train_hist[-1]:.4f} | val {val_hist[-1]:.4f}")
    return model, train_hist, val_hist, time.perf_counter() - t0

model, train_hist, val_hist, secs = train_model()
print(f"Training time: {secs:.1f}s")
```

```python
plt.figure(figsize=(6, 4))
plt.plot(train_hist, label="Training loss"); plt.plot(val_hist, label="Validation loss")
plt.xlabel("Epoch"); plt.ylabel("BCE loss"); plt.title("Loss curves"); plt.legend(); plt.show()

def evaluate(model, Xs, y_true, threshold=0.5):
    prob = predict_proba(model, Xs)
    pred = (prob >= threshold).astype(int)
    return {
        "accuracy":  accuracy_score(y_true, pred),
        "precision": precision_score(y_true, pred, zero_division=0),
        "recall":    recall_score(y_true, pred, zero_division=0),
        "f1":        f1_score(y_true, pred, zero_division=0),
        "roc_auc":   roc_auc_score(y_true, prob),
    }, pred

metrics, pred = evaluate(model, Xs_test, y_test)
print(pd.Series(metrics).round(4))

cm = confusion_matrix(y_test, pred, labels=[0, 1])
ConfusionMatrixDisplay(cm, display_labels=["Class 0", "Class 1"]).plot(cmap="Blues")
plt.title("Confusion matrix (test set)"); plt.show()
```

```python
def plot_boundary(model, X_raw, y_raw, resolution=300, point=None, show_prob=False, title="Learned decision boundary"):
    pad = 0.5
    xx, yy = np.meshgrid(np.linspace(X[:, 0].min() - pad, X[:, 0].max() + pad, resolution),
                         np.linspace(X[:, 1].min() - pad, X[:, 1].max() + pad, resolution))
    grid = scaler.transform(np.c_[xx.ravel(), yy.ravel()]).astype(np.float32)   # same scaler
    probs = predict_proba(model, grid).reshape(xx.shape)                         # from the model

    plt.figure(figsize=(7, 5))
    if show_prob:
        cs = plt.contourf(xx, yy, probs, levels=20, cmap="RdBu_r", alpha=0.6)
        plt.colorbar(cs, label="P(Class 1)")
    else:
        plt.contourf(xx, yy, (probs >= 0.5).astype(float), levels=[-0.5, 0.5, 1.5],
                     colors=["#cfe0f7", "#fbdcb5"])
    plt.contour(xx, yy, probs, levels=[0.5], colors="navy", linewidths=2)
    plt.scatter(X_raw[y_raw == 0, 0], X_raw[y_raw == 0, 1], s=18, c="#1f4e9c", edgecolors="white", label="Class 0 (test)")
    plt.scatter(X_raw[y_raw == 1, 0], X_raw[y_raw == 1, 1], s=18, c="#e07b00", edgecolors="white", label="Class 1 (test)")
    if point is not None:
        plt.scatter(*point, marker="*", s=300, c="red", edgecolors="white", label="Your point")
    plt.xlabel("x1"); plt.ylabel("x2"); plt.title(title); plt.legend(); plt.show()

plot_boundary(model, X_test, y_test)
plot_boundary(model, X_test, y_test, show_prob=True, title="Probability of Class 1")
```

```python
configs = {"1 layer x 8": [8], "2 layers x 16": [16, 16], "3 layers x 32": [32, 32, 32]}
rows = []
for name, layers in configs.items():
    m, tl, vl, t = train_model(layers, verbose=False)
    met, _ = evaluate(m, Xs_test, y_test)
    rows.append({"architecture": name, **met, "final_val_loss": vl[-1], "train_time_s": t})
results = pd.DataFrame(rows).round(4)
display(results)

results.set_index("architecture")[["accuracy", "f1", "roc_auc"]].plot.bar(figsize=(7, 4), ylim=(0, 1.05))
plt.title("Test metrics by architecture"); plt.xticks(rotation=0); plt.show()
```

```python
for lr in [0.001, 0.01, 0.1]:
    m, tl, vl, t = train_model([16, 16], lr=lr, epochs=100, verbose=False)
    met, _ = evaluate(m, Xs_test, y_test)
    print(f"lr={lr}: accuracy={met['accuracy']:.3f}, f1={met['f1']:.3f}, val_loss={vl[-1]:.3f}")
    plot_boundary(m, X_test, y_test, resolution=150, title=f"Boundary, lr={lr}")
```

```python
def predict_point(x1, x2, threshold=0.5):
    scaled = scaler.transform(np.array([[x1, x2]], dtype=np.float32)).astype(np.float32)
    p = float(predict_proba(model, scaled)[0])
    print(f"P(Class 1) = {p:.3f} -> predicted Class {int(p >= threshold)}")
    plot_boundary(model, X_test, y_test, point=(x1, x2))

predict_point(0.5, 0.25)
```

```python
import ipywidgets as w
from IPython.display import clear_output

x1s = w.FloatSlider(value=0.5, min=-1.5, max=2.5, step=0.05, description="x1")
x2s = w.FloatSlider(value=0.25, min=-1.0, max=1.5, step=0.05, description="x2")
btn = w.Button(description="Predict Class", button_style="primary")
out = w.Output()

def on_click(_):
    with out:
        clear_output(wait=True)
        predict_point(x1s.value, x2s.value)

btn.on_click(on_click)
display(x1s, x2s, btn, out)
```
